```python
#Verificando as informações de cada variavel

def profile_features(df: pd.DataFrame) -> pd.DataFrame:
    """
    Retorna um DataFrame com métricas descritivas para cada coluna.
    """
    info = {
        "feature":        [],
        "Tipo":           [],
        "Categórico":    [],
        "Binário":        [],
        "Qtd var unico":  [],
        "Min":            [],
        "Max":            [],
        "% Qtd de Nulos": [],
    }

    for col in df.columns:
        serie   = df[col]
        dtype   = serie.dtype
        n_uniq  = serie.nunique(dropna=True)
        is_cat  = pd.api.types.is_object_dtype(dtype) or isinstance(dtype, pd.CategoricalDtype)
        is_num  = pd.api.types.is_numeric_dtype(dtype)

        info["feature"].append(col)
        info["Tipo"].append(str(dtype))
        info["Categórico"].append(is_cat)
        info["Binário"].append(n_uniq == 2)
        info["Qtd var unico"].append(n_uniq)
        info["Min"].append(serie.min() if is_num else None)
        info["Max"].append(serie.max() if is_num else None)
        info["% Qtd de Nulos"].append(
            round(serie.isna().mean() * 100, 2)
        )

    return pd.DataFrame(info)


# ── Uso ─────────────────────────────────────────────────────────
result = profile_features(dados_selecionados)
print(result.to_string(index=False))
```

```python
#Filtrando as variaveis para a geração do modelo
variaveis2 = ['target_default', 'credit_limit', 'income', 'n_defaulted_loans', 'n_accounts', 'n_bankruptcies', 'n_issues', 'external_data_provider_fraud_score', 'reported_income']
dados_selecionados2 = dados[variaveis2]
```

```python
#Substituindo os valores nulos pela mediana
dados_selecionados2['credit_limit'] = dados_selecionados2['credit_limit'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['credit_limit'].median())
dados_selecionados2['income'] = dados_selecionados2['income'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['income'].median())
dados_selecionados2['n_defaulted_loans'] = dados_selecionados2['n_defaulted_loans'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['n_defaulted_loans'].median())
dados_selecionados2['n_accounts'] = dados_selecionados2['n_accounts'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['n_accounts'].median())
dados_selecionados2['n_bankruptcies'] = dados_selecionados2['n_bankruptcies'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['n_bankruptcies'].median())
dados_selecionados2['n_issues'] = dados_selecionados2['n_issues'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['n_issues'].median())
dados_selecionados2['external_data_provider_fraud_score'] = dados_selecionados2['external_data_provider_fraud_score'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['external_data_provider_fraud_score'].median())
dados_selecionados2['reported_income'] = dados_selecionados2['reported_income'].replace([np.inf, -np.inf], np.nan).fillna(dados_selecionados2['reported_income'].median())
```

```python
#Filtrando as linhas não nulas da variavel alvo
dados_selecionados2['target_default'] = dados_selecionados2['target_default'].astype(int)
dados_selecionados2 = dados_selecionados2.dropna(subset=['target_default'])

result = profile_features(dados_selecionados2)
print(result.to_string(index=False))
```

```python
#Organizando os dados para o modelo (Teste e Treino)
class_names = [str(cls) for cls in sorted(dados_selecionados2['target_default'].unique())]

X_dados = dados_selecionados2[dados_selecionados2.columns.drop('target_default')].values
y_dados = dados_selecionados2['target_default'].values

X_train, X_test, y_train, y_test = train_test_split(X_dados, y_dados, test_size=0.3, random_state=42, stratify=y_dados)
```

```python
#Cria e treina o modelo Random Forest
clf = RandomForestClassifier(
    n_estimators=100,       # número de árvores
    max_depth=None,         # profundidade máxima (None = sem limite)
    min_samples_split=2,    # mínimo de amostras para dividir um nó
    min_samples_leaf=1,     # mínimo de amostras em uma folha
    max_features="sqrt",    # features por divisão ("sqrt" é padrão para clf)
    random_state=42,
    n_jobs=-1               # usa todos os CPUs
)

clf.fit(X_train, y_train)
```

```python
#Previsões
y_pred = clf.predict(X_test)
y_prob = clf.predict_proba(X_test)
```

```python
# Métricas
print(f"\nAcurácia: {accuracy_score(y_test, y_pred):.4f}")
print(f"\nRelatório de Classificação:\n")
print(classification_report(y_test, y_pred, target_names=class_names))
```
